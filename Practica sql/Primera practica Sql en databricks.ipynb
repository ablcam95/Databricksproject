#Creacion de tabla con SQL en databricks

```python
%sql
CREATE TABLE IF NOT EXISTS workspace.default.users (
    id INT,
    name STRING,
    email STRING,
    age INT,
    create_at TIMESTAMP
)USING DELTA;
```

**USING DELTA = guarda la table en formato delta lake**

# Aplicar un insert de manera correcta
### Transaccion atomica
cada operacion es una transaccion completa y aislada (ACID).
- Aislamiento garantizado entre operaciones.
- Capacidad de revertir con time travel.
###Transaccion log (_delta_log)
cada transaccion genera un nuevo commit en el log.
- registro inmutable de todos los cambios.
- cada commit crea una nueva version de la tabla.
